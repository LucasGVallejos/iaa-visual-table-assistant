````python
import torch
import ultralytics
import cv2
import mlflow
import onnx
import onnxruntime
import dvc
import numpy as np
import pandas as pd

def check_version_and_validate(tool, current, expected):
  """
  Check if the current tool version is the same as the expected or raise an error.

  Useful to prevent weird errors related to breacking tool version changes
  """
  print(f"{tool}: {current}")

  if current != expected:
    raise Exception(f"Warning: Expected tool {tool}. Expected: {expected}, but got {current}")

  return current

check_version_and_validate("Torch", current=torch.__version__, expected="2.10.0+cpu")
check_version_and_validate("OpenCV", current=cv2.__version__, expected="4.13.0")
check_version_and_validate("ONNX Runtime", current=onnxruntime.__version__, expected="1.26.0")

is_cuda_avaiable = check_version_and_validate("CUDA available", current=torch.cuda.is_available(), expected=False)
if is_cuda_avaiable:
    print("CUDA device:", torch.cuda.get_device_name(0))
````
